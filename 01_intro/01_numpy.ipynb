```python
# Randomly generated arrays
np.random.rand(5,2)
```

```python
# To make the results reproducible, we use random seed
np.random.seed(2)
100 * np.random.rand(5, 2)
```

```python
# To draw numbers form the standard normal distribution, we add an n to the name rand
np.random.seed(2)
np.random.randn(5, 2)
```

```python
# Random integers
np.random.seed(2)
np.random.randint(low=0, high=100, size=(5, 2))
```
